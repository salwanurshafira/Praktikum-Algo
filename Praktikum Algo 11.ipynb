```python
"""
Nama : Salwa Nur Shafira
NIM : 065002500019
Tanggal : Kamis, 4 Desember 2025
Tugas : Sebuah kelas yang menerapkan method getter dan setter
"""

class Mahasiswa:
    def __init__(self, nama=None, nilai=None):
        self._nama = nama
        self._nilai = nilai

    def get_nama(self):
        return self._nama

    def get_nilai(self):
        return self._nilai

    def set_nama(self, nama_baru):
        self._nama = nama_baru

    def set_nilai(self, nilai_baru):
        self._nilai = nilai_baru

objek_mahasiswa = None
program_berjalan = True

def tampilkan_menu():
    print("\n==== Program OOP ====")
    print("1. Mendeklarasikan Objek")
    print("2. Menampilkan Objek")
    print("3. Mengubah Nilai Objek")
    print("4. Menghapus Objek")
    print("5. Keluar Dari Program")

def get_valid_integer_input(prompt):
    while True:
        input_str = input(prompt)
        if input_str.isdigit():
            return int(input_str)
        elif input_str == "":
            return None
        else:
            print("Input harus berupa angka. Coba lagi.")

def operasikan_program():
    global objek_mahasiswa
    global program_berjalan

    while program_berjalan:
        tampilkan_menu()

        pilihan_str = input("Masukkan Pilihan Berupa Angka (1/2/3/4/5): ")

        if pilihan_str not in ('1', '2', '3', '4', '5'):
            print("\nInput tidak valid. Masukkan angka 1 sampai 5.")
            continue

        pilihan = int(pilihan_str)

        if pilihan == 1:
            if objek_mahasiswa is not None:
                print("\nObjek sudah dideklarasikan. Silakan hapus objek yang ada terlebih dahulu (pilihan 4).")
                continue

            nama_input = input("Masukkan Namamu: ")

            nilai_awal = get_valid_integer_input("Masukkan Nilaimu: ")

            if nilai_awal is not None:
                objek_mahasiswa = Mahasiswa(nama_input, nilai_awal)
                print("Data Berhasil Ditambahkan.")
            else:
                 print("\nInput nilai tidak boleh kosong pada pilihan 1. Deklarasi dibatalkan.")

        elif pilihan == 2:
            if objek_mahasiswa is None:
                nama = None
                nilai = None

                print(f"Nama: {nama}")
                print(f"Nilai: {nilai}")
            else:
                nama = objek_mahasiswa.get_nama()
                nilai = objek_mahasiswa.get_nilai()

                print(f"Nama: {nama}")
                print(f"Nilai: {nilai}")

        elif pilihan == 3:
            if objek_mahasiswa is None:
                print("\nObjek belum dideklarasikan.")
            else:
                perubahan_terjadi = False
                _ = input("Apa yang ingin diubah (Nama/Nilai): ")
                nilai_baru = get_valid_integer_input("Masukkan Nilai: ")

                if nilai_baru is not None:
                    objek_mahasiswa.set_nilai(nilai_baru)
                    perubahan_terjadi = True

                if not perubahan_terjadi:
                     print("Tidak ada perubahan yang dilakukan.")
                else:
                     print("\nData nilai berhasil dirubah.")

        elif pilihan == 4:
            if objek_mahasiswa is None:
                print("\nObjek belum dideklarasikan.")
            else:
                objek_mahasiswa = None
                print("\nData Berhasil Dihapus.")

        elif pilihan == 5:
            program_berjalan = False
            print("\nTerima Kasih Sudah Menggunakan Program Saya")

operasikan_program()
```
